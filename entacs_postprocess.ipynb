```python
import json
import sys
import csv
```

```python
def json_to_csv(in_filepath, out_filepath, start_idx=0, end_idx=1000):
    """
    Process the JSON file which is output by the deepseek prompting
    save the codemixed text and monolingual translation for the given indices
    in_filepath: string, path to the json file of the model response
    out_filepath: string, path to output csv file
    start_idx: int, starting index of the sentences to be printed
    end_idx: int, ending index of sentences to be printed
    """

    try: 
        #load in json file
        with open(in_filepath, 'r', encoding='utf-8') as f: 
            data = json.load(f)
            print(f"num items in input: ", len(data))
        
        dict_list = []
        
        #process each object into dict
        for item in data[start_idx:end_idx]: 
            row = {"sentence id": item['timestamp'],
                   "codemixed text": item['original_text'],
                   "monolingual text": item['model_response']}
            dict_list.append(row)

        #write to output file
        with open(out_filepath, 'w', newline='') as f:
            FIELDNAMES = ['sentence id', 'codemixed text', 'monolingual text']
            writer = csv.DictWriter(f, fieldnames=FIELDNAMES)
            writer.writeheader()
            writer.writerows(dict_list)

        print(f"Processing complete, {len(dict_list)} sentences processed, written to {out_filepath}")
            

    except Exception as e:
        print(f"Mama an error behind you: {str(e)}")
        sys.exit(1)
```

```python
json_to_csv("ENTACS_sentiment_translated.json", "translated.csv")
```
